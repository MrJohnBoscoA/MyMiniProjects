## Titanic Dataset brief-Analysis 

```python
# Importing Libraries

import numpy as np
import pandas as pd

# Loading the Dataset ( train.xlxs is the titanic dataset)

data = pd.read_excel('train.xlsx')

# Exploreing the Dataset

print(data.head())  # display First 5 rows
print(data.shape)   # Number of rows and columns
print(data.dtypes)  # Data types of each column
```

```python
# Data pre-processing

data.drop('Cabin', axis=1, inplace=True)  # Drop 'Cabin' coloumn
data["Age"] = data["Age"].fillna(data["Age"].mean()) # Fill missing 'Age' with mean
data.dropna(subset=['Embarked'], inplace=True)  # Drop rows with missing 'Embarked' values

```

```python
# Data Analysis

overall_survival_rate = data['Survived'].mean()
male_survival_rate = data[data['Sex'] == 'male']['Survived'].mean()
female_survival_rate = data[data['Sex'] == 'female']['Survived'].mean()

avg_age_male = data[data['Sex'] == 'male']['Age'].mean()
avg_age_female = data[data['Sex'] == 'female']['Age'].mean()

survival_by_class = data.groupby('Pclass')['Survived'].mean()
survival_by_embark = data.groupby('Embarked')['Survived'].mean()

#  Data Manipulation

crosstab = pd.crosstab(data['Pclass'], data['Sex'], values=data['Survived'], aggfunc='sum')
stacked = crosstab.stack()
unstacked = stacked.unstack()
print(crosstab)

# Mergeing with original dataset (based on 'Pclass')

merged = data.merge(crosstab, on='Pclass', how='left')

# Grouping by 'Pclass' and calculate average age

avg_age_by_class = data.groupby('Pclass')['Age'].mean()

# Group by 'Pclass' and compare survival stats

survival_stats_by_class = data.groupby('Pclass')['Survived'].agg(['min', 'max', 'mean'])

# Filter passengers older than 30

older_than_30 = data[data['Age'] > 30]

# Data Transformation

data['Child'] = data['Age'].apply(lambda x: 'Yes' if x < 18 else 'No')
data['Embarked'] = data['Embarked'].map({'S': 'S', 'C': 'C', 'Q': 'Q'})
data['Name_Length'] = data['Name'].apply(len)
```

```python
# Survival Rates
print(f"Overall Survival Rate: {overall_survival_rate:.2%}")
print(f"Male Survival Rate: {male_survival_rate:.2%}")
print(f"Female Survival Rate: {female_survival_rate:.2%}")

# Average Ages
print(f"Average Age of Male Passengers: {avg_age_male:.2f}")
print(f"Average Age of Female Passengers: {avg_age_female:.2f}")

# Survival by Class
print("\nSurvival Rate by Passenger Class:")
print(survival_by_class)

# Survival by Embarkation Point
print("\nSurvival Rate by Embarkation Point:")
print(survival_by_embark)

# Crosstab Summary
print("\nSurvival Count Crosstab (Pclass vs Sex):")
print(crosstab)

# Average Age by Class
print("\nAverage Age by Passenger Class:")
print(avg_age_by_class)

# Survival Stats by Class
print("\nSurvival Statistics by Passenger Class (min, max, mean):")
print(survival_stats_by_class)

# Passengers Older Than 30
print(f"\nNumber of Passengers Older Than 30: {older_than_30.shape[0]}")

# Feature Engineering Summary
print("\nSample of 'Child' Feature (Age < 18):")
print(data['Child'].value_counts())

print("\nSample of 'Name_Length' Feature:")
print(data['Name_Length'].describe())
```
