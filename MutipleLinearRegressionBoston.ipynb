# Creating a Multiple Linear regression Model

```python
# importing required libraries 

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# importing a csv file with the boston_house_prices dataset 

df = pd.read_csv('boston_house_prices.csv')
```

```python
df
```

```python
df.describe()
```

```python
# Checking for null values in the dataset

df.isnull().sum()
```

```python
# assiging the feature and target varibles to the array X,y

X = df[['CRIM','ZN','INDUS','CHAS','NOX','RM','AGE','DIS','RAD','TAX','PTRATIO','B','LSTAT']]
y = df['MEDV']
```

```python
X
```

```python
y
```

```python
# importing sklearn for train test split

from sklearn.model_selection import train_test_split 
```

```python
# splittinig the dataset into train and test here we are spliting it train 80% and test as 20% , and random state to 3

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=3)
```

```python
X_train
```

```python
X_test
```

```python
# importing sklearn for Standard Scaler ( we r using this to Standardization the dataset)

from sklearn.preprocessing import StandardScaler
```

```python
# assigning the StandardScaler to varible scaler

scaler = StandardScaler()
```

```python
# standardizting the train and test dataset by scaler.fit_trasform  and storing the standardized data to varibles X_train_scaled,X_test_scaled respectively

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.fit_transform(X_test)
```

```python
X_train_scaled
```

```python
# importing sklearn for linear regression 

from sklearn.linear_model import LinearRegression
```

```python
# creatng  the regression  model 

model = LinearRegression()
```

```python
# fitting the train set to the model

model.fit(X_train_scaled,y_train)
```

```python
# performing predictions on  the test data based on the train dataset

model_pred = model.predict(X_test_scaled)
```

```python
model_pred
```

```python
# comparing the acutal (y_test) and predicted values (model_pred) 

y_test
```

```python
# importing sklearn for mean squared error 

from sklearn.metrics import mean_squared_error
```

```python
# finding the mean_squared_error for the acutal and predicted values and storing it in mse

mse = mean_squared_error(y_test,model_pred)
```

```python
# the mean squared error is 16.94%  so the accuary of the model is 83.06%

mse
```

## Conclusion : mean squared error is 16.94%  so the accuary of the model is 83.06%
