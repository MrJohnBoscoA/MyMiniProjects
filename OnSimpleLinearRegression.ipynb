## Performing a simple Linear regression Model

```python
# creating a dictorniary with manual entries to test the model 

data = {
    'Area':[1600,1800,2400,3000,3200,3600,4000,4100,4200],
    'Bedroom':[3,4,3,5,4,3,5,4,3],
    'Bathrooms':[2,3,3,4,3,2,4,3,3],
    'Age':[10,5,8,4,6,12,3,7,6],
    'Price':[80,95,120,140,155,75,180,160,110]
}
```

```python
data
```

```python
# importing pandas to create the manual data into a dataframe

import pandas as pd
df = pd.DataFrame(data)
```

```python
df
```

```python
# importing sklearn for train test split and linear regression 

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
# assiging the feature and target varibles to the array X,y

X = df[['Area','Bedroom','Bathrooms','Age']]
y = df['Price']
```

```python
# splittinig the dataset into train and test here we are spliting it train 80% and test as 20% , and random state to 3

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=3)
```

```python
# creatng  the linear regression  model 

model = LinearRegression()
```

```python
# fitting the train set to the model

model.fit(X_train,y_train)
```

```python
# performing predictions on  the test data

y_pred = model.predict(X_test)
```

```python
# comparing the acutal and predicted values 

result = pd.DataFrame({'Actual' : y_test, 'Predicted':y_pred})
```

```python
result
```

```python
# calculating accuracy of the model

accuracy = model.score(X_train,y_train)
```

```python
accuracy
```

```python
# performing predictions for custom / new entry 

new_house = pd.DataFrame({
    'Area':[3300],
    'Bedroom':[4],
    'Bathrooms':[3],
    'Age':[6]
    
})
```

```python
predict_new_house = model.predict(new_house)
```

```python
predict_new_house
```

## Conclusion : the model has attained more than 80 % accuracy
