## Titanic Data Processing

```python
# importing libraries
import pandas as pd
import numpy as np

# Loading the Titanic data ( train.xlxs is the titanic dataset)
df = pd.read_excel('train.xlsx')

# 1. Binning 'Age' and 'Fare'
age_bins = [0, 12, 20, 60, np.inf]
age_labels = ['Child', 'Teen', 'Adult', 'Elderly']
df['AgeGroup'] = pd.cut(df['Age'], bins=age_bins, labels=age_labels)

fare_bins = [0, 30, 70, np.inf]
fare_labels = ['Low', 'Medium', 'High']
df['FareRange'] = pd.cut(df['Fare'], bins=fare_bins, labels=fare_labels)

print(df['AgeGroup'])
print(df['FareRange'])
```

```python
# 2. Aggregation ("Aggregation is about summarizing your data by groups.")

pclass_agg = df.groupby('Pclass')[['Age', 'Fare']].mean()
sex_agg = df.groupby('Sex').agg({'PassengerId':'count', 'Age':'mean'})
print(pclass_agg)
print(sex_agg)
```

```python
# 3. Decomposition of 'Name' (" splitting ")

df['Title'] = df['Name'].str.extract(r',\s*([^\.]*)\.', expand=False)
df['LastName'] = df['Name'].str.extract(r'^([^\s,]+)', expand=False)
print(df['Title'])
print(df['LastName'])
```

```python
# 4. Feature Creation

df['FamilySize'] = df['SibSp'] + df['Parch'] # sibling + parents
df['IsAlone'] = np.where(df['FamilySize'] == 0, 1, 0)
print(df['FamilySize'])
print(df['IsAlone'])
```

```python
# 5. Feature Transformation (Encoding)
df['Sex_Label'] = df['Sex'].map({'male':0, 'female':1}) # Label encoding
embarked_dummies = pd.get_dummies(df['Embarked'], prefix='Embarked') # one-hot encoding
df = pd.concat([df, embarked_dummies], axis=1)

# Display first few rows
print(df.head())
```

## Conclusion :

```python

- **Age and Fare Binning:**  
  The 'Age' and 'Fare' variables were successfully binned into categorical groups, improving interpretability for analysis and modeling.

- **Aggregated Insights:**  
  Group-wise summaries (by class and sex) reveal clear differences in average age, fare, and passenger count, illustrating social and economic patterns among Titanic passengers.

- **Name Decomposition:**  
  Extracting titles and last names from the 'Name' column allows for further grouping and may help identify social status or family connections.

- **New Feature Creation:**  
  The engineered features 'FamilySize' and 'IsAlone' provide more insight into passenger groupings, which could have influenced survival probabilities.

- **Feature Transformation:**  
  Categorical data (like 'Sex' and 'Embarked') was encoded numerically, making the dataset suitable for machine learning models.

- **Data Ready for Analysis:**  
  After data cleaning, aggregation, decomposition, and transformation, the dataset now contains meaningful and machine-friendly features, enabling improved model training and richer analyses.

***

You can further tailor these conclusions depending on the specific findings or visualizations in your notebook!

[1](http://localhost:8888/notebooks/DataCleaningassign2.ipynb#Titanic-Data-Cleaning)
```
