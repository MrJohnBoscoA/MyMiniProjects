## Creating classes and objects 

```python
class Rectangle:
    def __init__(self, length, width):
        # Initializing the length and width of the rectangle
        self.length = length
        self.width = width

    def calculate_area(self):
        # Area = length * width
        return self.length * self.width

    def calculate_perimeter(self):
        # Perimeter = 2 * (length + width)
        return 2 * (self.length + self.width)


# Using the class that we created 

# Test1
rect1 = Rectangle(5, 3)
print("Rect1 - Length:", rect1.length, "Width:", rect1.width)
print("Rect1 - Area:", rect1.calculate_area())            
print("Rect1 - Perimeter:", rect1.calculate_perimeter())  

# Test2
rect2 = Rectangle(10, 7)
print("\nRect2 - Length:", rect2.length, "Width:", rect2.width)
print("Rect2 - Area:", rect2.calculate_area())            # 70
print("Rect2 - Perimeter:", rect2.calculate_perimeter())  # 34

# Test3 
rect3 = Rectangle(4, 4)
print("\nRect3 - Length:", rect3.length, "Width:", rect3.width)
print("Rect3 - Area:", rect3.calculate_area())            # 16
print("Rect3 - Perimeter:", rect3.calculate_perimeter())  # 16
```

```python
class Employee:
    def __init__(self, name, emp_id, department, salary):
        self.name = name
        self.emp_id = emp_id
        self.department = department
        self.salary = salary

    def get_employee_details(self):
        print(f"Name: {self.name}")
        print(f"Employee ID: {self.emp_id}")
        print(f"Department: {self.department}")
        print(f"Salary: {self.salary}")

    def calculate_salary(self):
        return self.salary


# Example:
emp = Employee("John Bosco A", "E12345", "Engineering", 90000)

# Print details
emp.get_employee_details()

# Calculate salary
current_salary = emp.calculate_salary()
print(f"Calculated Salary: {current_salary}")
```

```python
class BankAccount:
    def __init__(self, acc_number, acc_holder, balance=0.0):
        self.acc_number = acc_number
        self.acc_holder = acc_holder
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            print("Deposit amount must be positive.")
            return
        self.balance += amount
        print(f"Deposited {amount:.2f}. New balance: {self.balance:.2f}")

    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal amount must be positive.")
            return
        if amount > self.balance:
            print("Insufficient funds.")
            return
        self.balance -= amount
        print(f"Withdrew {amount:.2f}. New balance: {self.balance:.2f}")

    def get_account_details(self):
        print("Account Details:")
        print(f"  Account Number: {self.acc_number}")
        print(f"  Account Holder: {self.acc_holder}")
        print(f"  Balance: {self.balance:.2f}")

# Testing
    # Create first account
acc1 = BankAccount(acc_number="123456789", acc_holder="John Bosco A", balance=1000.0)
acc1.get_account_details()
acc1.deposit(250.0)
acc1.withdraw(100.0)
acc1.get_account_details()

print("\n") 

    # Create second account with different details
acc2 = BankAccount(acc_number="987654321", acc_holder="johnson", balance=500.0)
acc2.get_account_details()
acc2.deposit(150.0)
acc2.withdraw(700.0)  # Will show insufficient funds
acc2.withdraw(200.0)
acc2.get_account_details()
```

```python
# Base class
class Animal:
    def make_sound(self):
        raise NotImplementedError("Subclasses should implement this!")

# Derived classes
class Dog(Animal):
    def make_sound(self):
        print("Bark")

class Cat(Animal):
    def make_sound(self):
        print("Meow")

class Lion(Animal):
    def make_sound(self):
        print("Roar")

# Created objects of each derived class
dog = Dog()
cat = Cat()
lion = Lion()

# Test the make_sound method
dog.make_sound()   
cat.make_sound()   
lion.make_sound()  
```

```python
class Student:
    def __init__(self, name, roll_number, age):
        self.__name = name
        self.__roll_number = roll_number
        self.__age = age

    def get_student_details(self):
        print("Student Details:")
        print(f"  Name        : {self.__name}")
        print(f"  Roll Number : {self.__roll_number}")
        print(f"  Age         : {self.__age}")

#Testing the class with different student objects
student1 = Student("Aarav Mehta", "S101", 20)
student2 = Student("Diya Sharma", "S102", 19)
student3 = Student("Rohan Iyer", "S103", 21)

student1.get_student_details()
print()
student2.get_student_details()
print()
student3.get_student_details()
```

```python
class Bank:
    def __init__(self, name, branch, ifsc):
        self.__name = name
        self.__branch = branch
        self.__ifsc = ifsc

    def get_bank_details(self):
        print("Bank Details:")
        print(f"  Bank Name : {self.__name}")
        print(f"  Branch    : {self.__branch}")
        print(f"  IFSC Code : {self.__ifsc}")

#Creating and testing Bank objects
bank1 = Bank("State Bank of India", "MG Road, Bengaluru", "SBIN0001234")
bank2 = Bank("HDFC Bank", "Jayanagar, Bengaluru", "HDFC0005678")
bank3 = Bank("ICICI Bank", "Indiranagar, Bengaluru", "ICIC0004321")

bank1.get_bank_details()
print()
bank2.get_bank_details()
print()
bank3.get_bank_details()
```

```python
class Product:
    def __init__(self, name, price, quantity):
        self.__name = name
        self.__price = price
        self.__quantity = quantity

    def calculate_total_cost(self):
        return self.__price * self.__quantity

    def get_product_details(self):
        print("Product Details:")
        print(f"  Name     : {self.__name}")
        print(f"  Price    : ₹{self.__price:.2f}")
        print(f"  Quantity : {self.__quantity}")
        print(f"  Total Cost: ₹{self.calculate_total_cost():.2f}")
        
# Creating product objects
product1 = Product("Wireless Mouse", 599.00, 2)
product2 = Product("Bluetooth Speaker", 1499.50, 1)
product3 = Product("USB-C Cable", 299.99, 3)

# Displaying product details
product1.get_product_details()
print()
product2.get_product_details()
print()
product3.get_product_details()
```
